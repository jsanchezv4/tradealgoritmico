```python
import yfinance as yf
import pandas_ta as ta
import plotly.graph_objects as go
from datetime import datetime, timedelta

class EstrategiaTrading:
    def __init__(self, simbolo, fecha_inicio, fecha_fin, intervalo='1m', dias=1):
        self.simbolo = simbolo
        self.fecha_inicio = fecha_inicio
        self.fecha_fin = fecha_fin
        self.intervalo = intervalo
        self.dias = dias
        self.capital_total = 20  # Capital total disponible para trading
        self.atr_length = None  # Longitud para el cálculo del ATR (se actualizará dinámicamente)
        self.atr_multiplier_stop_loss = 0.02  # Multiplicador para el cálculo del stop-loss basado en ATR
        self.atr_multiplier_take_profit = 0.03  # Multiplicador para el cálculo del take-profit basado en ATR
        self.porcentaje_max_posicion = 0.02  # Porcentaje máximo del capital total en una posición

    def descargar_datos(self):
        return yf.download(self.simbolo, start=self.fecha_inicio, end=self.fecha_fin, interval=self.intervalo)

    def identificar_tendencia(self, df):
        df['EMA_corta'] = ta.ema(df['Close'], length=20)
        df['EMA_larga'] = ta.ema(df['Close'], length=50)
        df['Tendencia'] = 'Indefinida'
        df.loc[df['EMA_corta'] > df['EMA_larga'], 'Tendencia'] = 'Alcista'
        df.loc[df['EMA_corta'] < df['EMA_larga'], 'Tendencia'] = 'Bajista'
        return df

    def calcular_rsi(self, df):
        df['RSI'] = ta.rsi(df['Close'], length=14)
        return df

    def evaluar_senales(self, df):
        if self.generar_senal_compra(df):
            self.gestionar_posicion(df, "comprar")
        elif self.generar_senal_venta(df):
            self.gestionar_posicion(df, "vender")
        else:
            print("No se detectaron señales de compra o venta.")

        self.detectar_evento_liquidez(df)

    def generar_senal_compra(self, df):
        return (df['EMA_corta'].iloc[-1] > df['EMA_larga'].iloc[-1]) and (df['RSI'].iloc[-1] < 30)

    def generar_senal_venta(self, df):
        return (df['EMA_corta'].iloc[-1] < df['EMA_larga'].iloc[-1]) and (df['RSI'].iloc[-1] > 70)

    def gestionar_posicion(self, df, accion):
        riesgo_por_operacion = self.capital_total * self.porcentaje_max_posicion

        # Verificar si self.atr_length es cero antes de realizar la división
        if self.atr_length == 0:
            print("Error: La longitud del ATR es cero. No se puede calcular el tamaño de la posición.")
            return

        # Calcular el tamaño de la posición en función del riesgo por operación y el ATR
        tamano_posicion = riesgo_por_operacion / self.atr_length

        if accion == "comprar":
            precio_compra = df.iloc[-1]['Close']
            stop_loss = precio_compra - (self.atr_multiplier_stop_loss * self.atr_length)
            take_profit = precio_compra + (self.atr_multiplier_take_profit * self.atr_length)
            print(f"Señal de compra detectada. Compra ahora a {precio_compra}, Stop-loss en {stop_loss}, Take-profit en {take_profit}")
            # Implementa la lógica para abrir una posición de compra

        elif accion == "vender":
            precio_venta = df.iloc[-1]['Close']
            stop_loss = precio_venta + (self.atr_multiplier_stop_loss * self.atr_length)
            take_profit = precio_venta - (self.atr_multiplier_take_profit * self.atr_length)
            print(f"Señal de venta detectada. Vende ahora a {precio_venta}, Stop-loss en {stop_loss}, Take-profit en {take_profit}")
            # Implementa la lógica para abrir una posición de venta

    def visualizar_datos(self, df):
        fig = go.Figure(data=[go.Candlestick(x=df.index,
                                            open=df['Open'],
                                            high=df['High'],
                                            low=df['Low'],
                                            close=df['Close'])])

        fig.add_trace(go.Scatter(x=df.index, y=df['EMA_corta'], mode='lines', name='EMA (Corta)', line=dict(color='blue')))
        fig.add_trace(go.Scatter(x=df.index, y=df['EMA_larga'], mode='lines', name='EMA (Larga)', line=dict(color='red')))

        niveles_fibonacci = self.calcular_niveles_fibonacci(df)
        for nivel, precio in niveles_fibonacci.items():
            fig.add_shape(
                type="line",
                x0=df.index[0],
                y0=precio,
                x1=df.index[-1],
                y1=precio,
                line=dict(color='green', width=1, dash='dash'),
                name=f'Fib {nivel}'
            )

        fig.update_layout(title='Tendencia del precio (EMA) con Retracement de Fibonacci',
                          xaxis_rangeslider_visible=False)
        fig.show()

    def estrategia_reversion_media(self, df, ventana):
        # Calcular la media móvil exponencial
        df['Media'] = ta.ema(df['Close'], length=ventana)

        # Calcular la desviación estándar
        df['Desviacion'] = ta.stdev(df['Close'], length=ventana)

        # Calcular las bandas de Bollinger con la media móvil exponencial
        df['Banda_Superior'] = df['Media'] + 2 * df['Desviacion']
        df['Banda_Inferior'] = df['Media'] - 2 * df['Desviacion']

        # Generar señales de compra y venta
        df['Señal_Compra'] = (df['Close'] < df['Banda_Inferior']).astype(int)
        df['Señal_Venta'] = (df['Close'] > df['Banda_Superior']).astype(int)

        return df

    def cruce_medias_moviles(self, df, corta, larga):
        df['corta'] = ta.ema(df['Close'], length=corta)
        df['larga'] = ta.ema(df['Close'], length=larga)
        df['Posicion'] = 0

        df.loc[df['corta'] > df['larga'], 'Posicion'] = 1
        df.loc[df['corta'] < df['larga'], 'Posicion'] = -1

        return df

    def calcular_bbands(self, df):
        # Calcular la media móvil simple de 20 periodos
        df['SMA'] = ta.sma(df['Close'], length=20)

        # Calcular la desviación estándar de 20 periodos
        df['STD'] = ta.stdev(df['Close'], length=20)

        # Calcular las bandas de Bollinger
        df['BB_UPPER'] = df['SMA'] + 2 * df['STD']
        df['BB_LOWER'] = df['SMA'] - 2 * df['STD']

        return df

    def calcular_atr(self, df):
        self.verificar_datos_completos(df)  # Verificar que los datos sean válidos
        self.atr_length = ta.atr(df['High'], df['Low'], df['Close'], length=14).iloc[-1]

    def verificar_datos_completos(self, df):
        if df[['High', 'Low', 'Close']].isnull().values.any():
            raise ValueError("Hay valores faltantes en las columnas 'High', 'Low' o 'Close' del DataFrame.")


    def detectar_evento_liquidez(self, df):
        self.calcular_atr(df)  # Calcula el ATR antes de utilizarlo

        # Comparar el rango verdadero promedio con el cierre anterior para detectar eventos de liquidez
        if df.iloc[-1]['Close'] > df.iloc[-2]['Close'] + self.atr_length:
            print("¡Se detectó un evento de liquidez!")
    def calcular_niveles_fibonacci(self, df):
        maximo = df['High'].max()
        minimo = df['Low'].min()
        niveles_fibonacci = {
            0: minimo,
            0.236: maximo - (0.236 * (maximo - minimo)),
            0.382: maximo - (0.382 * (maximo - minimo)),
            0.5: maximo - (0.5 * (maximo - minimo)),
            0.618: maximo - (0.618 * (maximo - minimo)),
            0.786: maximo - (0.786 * (maximo - minimo)),
            1: maximo
        }
        return niveles_fibonacci


    def simular_trading_retroactivo(self):
        # Descargar datos históricos
        df = self.descargar_datos()

        # Identificar tendencia
        df = self.identificar_tendencia(df)

        # Calcular indicadores EMA, BBANDS y RSI
        df = self.calcular_bbands(df)
        df = self.calcular_rsi(df)

        # Calcular ATR y detectar eventos de liquidez
        self.detectar_evento_liquidez(df)

        # Evaluar señales de compra o venta
        self.evaluar_senales(df)

        # Visualizar patrones de velas y señales con Retracement de Fibonacci
        self.visualizar_datos(df)

        # Implementar estrategia de cruce de medias móviles
        df = self.cruce_medias_moviles(df, 20, 50)

        # Implementar estrategia de reversión a la media
        df = self.estrategia_reversion_media(df, ventana=20)


# Configuración de parámetros
simbolo = "BTC-USD" #@param ["BTC-USD", "ETH-USD", "ADA-USD", "SOL-USD", "DOT-USD", "USDT-USD", "DAI-USD", "USDC-USD", "LUNA-USD", "MATIC-USD", "AVAX-USD", "BNB-USD", "MANA-USD", "SAND-USD", "SLP-USD", "AXS-USD", "UNI-USD", "ALGO-USD", "AAVE-USD", "ENS-USD", "NEAR-USD", "FTM-USD", "PAXG-USD", "DOGE-USD", "TRX-USD", "ARB-USD", "LTC-USD", "XLM-USD", "OP-USD", "TRX-USD", "LUNA-USD"]
dias = "1"  #@param ["59", "15", "7", "1"]
intervalo = "15m" #@param ["1m", "5m", "15m", "1d"]

fecha_inicio = datetime.now() - timedelta(days=int(dias))
fecha_fin = datetime.now()

# Crear instancia de la estrategia de trading
estrategia = EstrategiaTrading(simbolo, fecha_inicio, fecha_fin, intervalo, dias)

# Realizar simulación retroactiva
estrategia.simular_trading_retroactivo()
```

```python
import yfinance as yf
import pandas_ta as ta
import plotly.graph_objects as go
from datetime import datetime, timedelta

class EstrategiaTrading:
    def __init__(self, simbolo, fecha_inicio, fecha_fin):
        self.simbolo = simbolo
        self.fecha_inicio = fecha_inicio
        self.fecha_fin = fecha_fin
        self.capital_total = 20  # Capital total disponible para trading

        # Llamar a la simulación retroactiva al inicializar la estrategia
        self.simular_trading_retroactivo()

    def descargar_datos(self):
        return yf.download(self.simbolo, start=self.fecha_inicio, end=self.fecha_fin, interval='1d')


    def encontrar_soportes_resistencias(self, df):
        if df is None:
            print("El DataFrame está vacío.")
            return

        # Implementación preliminar utilizando rolling windows para suavizar los datos y buscar mínimos y máximos locales
        if 'Low' in df.columns:
            df['Rolling_Min'] = df['Low'].rolling(window=20).min()
        else:
            print("No se pudo encontrar la columna 'Low' en el DataFrame.")
            return

        if 'High' in df.columns:
            df['Rolling_Max'] = df['High'].rolling(window=20).max()
        else:
            print("No se pudo encontrar la columna 'High' en el DataFrame.")
            return

        # Calcular niveles de soporte y resistencia basados en los mínimos y máximos suavizados
        soportes = df[df['Low'] == df['Rolling_Min']]['Low']
        resistencias = df[df['High'] == df['Rolling_Max']]['High']

        # Mostrar los niveles de soporte y resistencia encontrados
        print("Niveles de Soporte:")
        print(soportes)
        print("\nNiveles de Resistencia:")
        print(resistencias)


    def evaluar_volatilidad(self, df):
        # Implementación preliminar utilizando el Average True Range (ATR) como indicador de volatilidad
        atr = ta.atr(df['High'], df['Low'], df['Close'], length=14)  # Calcular el ATR con una ventana de 14 períodos
        ultimo_atr = atr.iloc[-1]  # Último valor del ATR

        # Evaluar la volatilidad basada en el último valor del ATR
        if ultimo_atr > 0.05:
            return "Alta"
        elif ultimo_atr < 0.02:
            return "Baja"
        else:
            return "Moderada"


    def visualizar_datos(self, df):
        # Implementación preliminar utilizando un gráfico interactivo de Plotly para visualizar los precios históricos y las bandas de Bollinger
        fig = go.Figure()

        # Agregar velas japonesas al gráfico
        fig.add_trace(go.Candlestick(x=df.index,
                                      open=df['Open'],
                                      high=df['High'],
                                      low=df['Low'],
                                      close=df['Close'], name='Precio Histórico'))

        # Agregar bandas de Bollinger al gráfico
        fig.add_trace(go.Scatter(x=df.index, y=df['BBANDS_Superior'], mode='lines', name='BBANDS Superior', line=dict(color='blue')))
        fig.add_trace(go.Scatter(x=df.index, y=df['BBANDS_Medio'], mode='lines', name='BBANDS Medio', line=dict(color='black')))
        fig.add_trace(go.Scatter(x=df.index, y=df['BBANDS_Inferior'], mode='lines', name='BBANDS Inferior', line=dict(color='red')))

        # Actualizar diseño del gráfico
        fig.update_layout(title='Precio Histórico con Bandas de Bollinger',
                          xaxis_rangeslider_visible=False)

        # Mostrar el gráfico
        fig.show()

    def calcular_bbands(self, df):
        bbands = ta.bbands(df['Close'], length=20, std=2)  # Bandas de Bollinger con una longitud de 20 y un multiplicador de 2 para la desviación estándar

        if bbands is not None:
            df['BBANDS_Superior'] = bbands['BBU_20_2.0']
            df['BBANDS_Medio'] = bbands['BBM_20_2.0']
            df['BBANDS_Inferior'] = bbands['BBL_20_2.0']
        else:
            print("No se pudieron calcular las bandas de Bollinger.")
            return None  # Devolver None si las bandas de Bollinger no se pueden calcular

        return df

    def tomar_decision_compra_venta(self, df):
        if 'BBANDS_Superior' not in df.columns:
            print("Las bandas de Bollinger no están disponibles.")
            return None

        ultimo_cierre = df['Close'].iloc[-1]
        banda_superior = df['BBANDS_Superior'].iloc[-1]
        banda_inferior = df['BBANDS_Inferior'].iloc[-1]
        volatilidad = self.evaluar_volatilidad(df)

        # Comprobamos si el precio está por encima de la banda superior de Bollinger
        if ultimo_cierre > banda_superior:
            return "Vender"  # Indicamos que se debería vender, ya que el precio podría revertirse a la baja

        # Comprobamos si el precio está por debajo de la banda inferior de Bollinger
        elif ultimo_cierre < banda_inferior:
            return "Comprar"  # Indicamos que se debería comprar, ya que el precio podría revertirse al alza

        # Si no se cumple ninguna de las condiciones anteriores, evaluamos la volatilidad para tomar una decisión más conservadora
        elif volatilidad == "Alta":
            return "Esperar"  # Indicamos que se debería esperar debido a la alta volatilidad del mercado

        else:
            return "Sin señal clara"  # No se encuentra ninguna señal clara para comprar o vender

    def simular_trading_retroactivo(self):
        df = self.descargar_datos()
        if not df.empty:
            df = self.calcular_bbands(df)
            self.encontrar_soportes_resistencias(df)
            self.visualizar_datos(df)
            decision = self.tomar_decision_compra_venta(df)
            print("Decisión:", decision)
            print("Analizar los niveles de soporte y resistencia para tomar decisiones de trading.")
        else:
            print("No se pueden realizar simulaciones porque el DataFrame está vacío.")

# Solicitar al usuario que ingrese los parámetros de entrada
simbolo = "BTC-USD" #@param ["BTC-USD", "ETH-USD", "ADA-USD", "SOL-USD", "DOT-USD", "USDT-USD", "DAI-USD", "USDC-USD", "LUNA-USD", "MATIC-USD", "AVAX-USD", "BNB-USD", "MANA-USD", "SAND-USD", "SLP-USD", "AXS-USD", "UNI-USD", "ALGO-USD", "AAVE-USD", "ENS-USD", "NEAR-USD", "FTM-USD", "PAXG-USD", "DOGE-USD", "TRX-USD", "ARB-USD", "LTC-USD", "XLM-USD", "OP-USD", "TRX-USD", "LUNA-USD"]
dias = "15"  #@param ["59", "15", "7", "1"]
intervalo = "15m" #@param ["1m", "5m", "15m", "1d"]

fecha_inicio = datetime.now() - timedelta(days=int(dias))
fecha_fin = datetime.now()

# Crear instancia de la estrategia de trading con los parámetros ingresados por el usuario
estrategia = EstrategiaTrading(simbolo, fecha_inicio, fecha_fin)
```
